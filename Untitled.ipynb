```python
import json
from pathlib import Path

IN  = Path("training_samples.geojson")            # your mixed file
OUT = Path("training_samples_fixed.geojson")      # cleaned output

def normalize_polygon(coords):
    """Ensure outer LinearRing is closed (first == last)."""
    if not coords:
        return coords
    ring = coords[0]
    if ring[0] != ring[-1]:
        ring = ring + [ring[0]]
    return [ring]

def read_mixed(path: Path):
    """Read either a single JSON or newline-delimited JSON (NDJSON)."""
    text = path.read_text(encoding="utf-8").strip()
    try:
        return [json.loads(text)]
    except json.JSONDecodeError:
        objs = []
        for line in text.splitlines():
            line = line.strip()
            if not line:
                continue
            try:
                objs.append(json.loads(line))
            except json.JSONDecodeError:
                pass
        return objs

objs = read_mixed(IN)

# Collect features from FeatureCollection + loose Feature lines
features = []
for obj in objs:
    if not isinstance(obj, dict):
        continue
    t = obj.get("type")
    if t == "FeatureCollection":
        for f in obj.get("features", []):
            if isinstance(f, dict) and f.get("type") == "Feature":
                features.append(f)
    elif t == "Feature":
        features.append(obj)

# Clean: drop non-GeoJSON keys in geometry (e.g., 'geodesic'), close rings, keep 'label'
cleaned = []
for f in features:
    geom = (f.get("geometry") or {}).copy()
    gtype = geom.get("type")
    coords = geom.get("coordinates")

    # keep only valid keys
    geom = {"type": gtype, "coordinates": coords}

    # close polygon rings
    if gtype == "Polygon" and coords:
        geom["coordinates"] = normalize_polygon(coords)

    props = f.get("properties", {}) or {}
    cleaned.append({"type": "Feature", "properties": props, "geometry": geom})

fixed = {"type": "FeatureCollection", "name": "training_samples_fixed", "features": cleaned}

OUT.write_text(json.dumps(fixed, ensure_ascii=False), encoding="utf-8")
print(f"✅ Wrote {len(cleaned)} features to {OUT}")
```

```python
# ---- Sentinel-2 (cloud-masked) ----
def s2_mask(img):
    qa = img.select('QA60')
    mask = qa.bitwiseAnd(1<<10).eq(0).And(qa.bitwiseAnd(1<<11).eq(0))
    return img.updateMask(mask).divide(10000)

s2 = (ee.ImageCollection('COPERNICUS/S2_SR')
      .filterDate(START, END)
      .filterBounds(samples_fc.geometry())
      .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 30))
      .map(s2_mask)
      .median())

ndvi = s2.normalizedDifference(['B8','B4']).rename('NDVI')
evi  = s2.expression(
    '2.5*((NIR-RED)/(NIR+6*RED-7.5*BLUE+1))',
    {'NIR': s2.select('B8'), 'RED': s2.select('B4'), 'BLUE': s2.select('B2')}
).rename('EVI')

s2_feats = s2.select(['B2','B3','B4','B8','B11','B12']).addBands([ndvi, evi])

# ---- Sentinel-1 (VV, VH, VV/VH) ----
s1 = (ee.ImageCollection('COPERNICUS/S1_GRD')
      .filterDate(START, END)
      .filterBounds(samples_fc.geometry())
      .filter(ee.Filter.eq('instrumentMode', 'IW'))
      .filter(ee.Filter.eq('resolution_meters', 10))
      .select(['VV','VH'])
      .median())

vv_vh = s1.select('VV').divide(s1.select('VH')).rename('VV_VH')

# ---- Combined stack ----
stack = s2_feats.addBands(s1).addBands(vv_vh)
print("✅ Bands:", stack.bandNames().getInfo())
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------
# Replace this confusion matrix with your actual matrix if different
# Rows = true, Cols = predicted
cm = np.array([
    [1376, 11, 0],   # corn
    [25, 837, 17],   # rice
    [0, 6, 1140]     # sugarcane
])

# ---------------------------
# Compute overall accuracy (po)
total = cm.sum()
po = np.trace(cm) / total   # observed agreement (Overall Accuracy)

# ---------------------------
# Compute Cohen's Kappa
# Expected agreement pe = sum_i (row_sum_i * col_sum_i) / total^2
row_sums = cm.sum(axis=1)
col_sums = cm.sum(axis=0)
pe = (row_sums * col_sums).sum() / (total ** 2)

kappa = (po - pe) / (1 - pe)

# ---------------------------
# Optional: Per-class metrics (for context)
precision = np.diag(cm) / col_sums   # TP / (TP+FP)
recall = np.diag(cm) / row_sums      # TP / (TP+FN)
f1 = 2 * precision * recall / (precision + recall)

# Print values
print("Overall Accuracy (OA): {:.4f} ({:.2f}%)".format(po, po*100))
print("Cohen's Kappa: {:.4f}".format(kappa))
print("\nPer-class (precision, recall, f1):")
for i, cls in enumerate(['corn','rice','sugarcane']):
    print(f" {cls:9s} -> precision: {precision[i]:.4f}, recall: {recall[i]:.4f}, f1: {f1[i]:.4f}")

# ---------------------------
# Plot OA and Kappa as a bar chart
labels = ['Overall Accuracy', "Cohen's Kappa"]
values = [po, kappa]

fig, ax = plt.subplots(figsize=(6,4))
bars = ax.bar(labels, values, color=['#FF9F1C', '#2EC4B6'])
ax.set_ylim(0, 1.05)
ax.set_ylabel('Score')
ax.set_title("Overall Accuracy and Cohen's Kappa")

# annotate bars with percentage (or decimal)
for i, v in enumerate(values):
    if i == 0:
        txt = f"{v*100:.2f}%"
    else:
        txt = f"{v:.4f}"
    ax.text(i, v + 0.03, txt, ha='center', fontsize=10)

plt.grid(axis='y', linestyle='--', linewidth=0.4)
plt.tight_layout()
# plt.savefig("oa_kappa.png", dpi=300)   # uncomment to save
plt.show()
```
