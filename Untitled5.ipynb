```python
import ee
import geemap

ee.Initialize(project='crop-classification-476512')

# ---- PARAMETERS ----
# AOI: if you already have a large AOI geometry, use it; otherwise set a bounding box
aoi = ee.Geometry.Rectangle([68.0, 6.5, 97.5, 37.5])  # India rough bbox (change if needed)

date_start = '2024-06-01'
date_end   = '2024-11-30'   # one crop season (Kharif example)
s2_scale = 10               # meters
s1_scale = 10

# class mapping (change or add classes as needed)
label_to_index = {'rice': 1, 'sugarcane': 2, 'corn': 3}
label_dict = ee.Dictionary(label_to_index)
```

```python
def mask_s2_clouds(img):
    # use QA60 band cloud mask (bits 10 and 11)
    qa = img.select('QA60')
    cloudBitMask = 1 << 10
    cirrusBitMask = 1 << 11
    mask = qa.bitwiseAnd(cloudBitMask).eq(0).And(qa.bitwiseAnd(cirrusBitMask).eq(0))
    return img.updateMask(mask).copyProperties(img, ['system:time_start'])

def add_s2_indices(img):
    # compute NDVI and EVI
    ndvi = img.normalizedDifference(['B8', 'B4']).rename('NDVI')
    # EVI formula: 2.5 * (NIR-Red)/(NIR + 6*Red - 7.5*Blue + 1) (approx)
    evi = img.expression(
        '2.5 * (NIR - RED) / (NIR + 6*RED - 7.5*BLUE + 1)',
        {'NIR': img.select('B8'), 'RED': img.select('B4'), 'BLUE': img.select('B2')}
    ).rename('EVI')
    return img.addBands([ndvi, evi])

s2 = (
    ee.ImageCollection('COPERNICUS/S2_SR')
      .filterDate(date_start, date_end)
      .filterBounds(aoi)
      .map(mask_s2_clouds)
      .map(add_s2_indices)
)

# pick median composite (keeps bands plus NDVI, EVI)
s2_median = s2.select(['B2','B3','B4','B8','B11','B12','NDVI','EVI']).median().clip(aoi)
# rename to friendly names
s2_bands = ['blue','green','red','nir','swir1','swir2','ndvi','evi']
s2_median = s2_median.rename(s2_bands)
```

```python
def s1_prep(img):
    # select VV and VH, convert to float, keep backscatter as-is
    img = img.select(['VV','VH']).toFloat()
    return img.copyProperties(img, ['system:time_start'])

s1 = (
    ee.ImageCollection('COPERNICUS/S1_GRD')
      .filter(ee.Filter.eq('instrumentMode', 'IW'))
      .filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV'))
      .filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VH'))
      .filterDate(date_start, date_end)
      .filterBounds(aoi)
      .map(s1_prep)
)

s1_median = s1.median().clip(aoi)
# add ratio
vv = s1_median.select('VV')
vh = s1_median.select('VH')
ratio = vv.divide(vh).rename('vv_vh_ratio')
s1_median = s1_median.addBands(ratio)
```

```python
# combine S2 and S1 stacks
stack = s2_median.addBands(s1_median.select(['VV','VH','vv_vh_ratio']))

print('stack band names:', stack.bandNames().getInfo())
# Example output: ['blue','green','red','nir','swir1','swir2','ndvi','evi','VV','VH','vv_vh_ratio']
```

```python
# mask for COPERNICUS/S2_SR images
def mask_s2_sr(img):
    # mask based on cloud probability and scene classification
    # Keep pixels where cloud probability < 50%
    cloudProb = img.select('MSK_CLDPRB')
    cloud_mask = cloudProb.lt(50)

    # SCL codes to exclude (cloud, cloud shadows, snow, etc):
    # SCL values: 0-11 (see docs). Exclude 3 (cloud shadow), 8 (cloud medium prob), 9 (cloud high prob),
    # 10 (thin cirrus), 11 (snow). Adjust as needed.
    scl = img.select('SCL')
    scl_good = scl.neq(3).And(scl.neq(8)).And(scl.neq(9)).And(scl.neq(10)).And(scl.neq(11))

    mask = cloud_mask.And(scl_good)

    # apply mask and preserve time property
    return img.updateMask(mask).copyProperties(img, ['system:time_start'])
```
